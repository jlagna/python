```python
def calculator(a, b):
    def add(): return a + b
    def sub(): return a - b
    def mul(): return a * b
    def div(): return a / b if b != 0 else "Divide by zero"
    return add(), sub(), mul(), div()

print(calculator(10, 5))
```

```python
def text_processor(sentence):
    def clean(text):
        return text.lower().replace(".", "")
    def count_words(text):
        return len(text.split())

    cleaned = clean(sentence)
    return cleaned, count_words(cleaned)

print(text_processor("Hello, Python World."))
```

```python
def text_processor(sentence):
    def clean(text):
        return text.lower().replace(".", "")
    def count_words(text):
        return len(text.split(","))

    cleaned = clean(sentence)
    return cleaned, count_words(cleaned)

print(text_processor("Hello, Python World."))
```

```python
def power_func(exp):
   def inner(base):
     return base ** exp
   return inner

square = power_func(2)
cube = power_func(3)

print(square(5),cube(2))
```

```python
def voting_system():
    votes = {"A": 0, "B": 0}

    def vote(candidate):
        if candidate in votes:
            votes[candidate] += 1
        return votes

    return vote

vote = voting_system()
print(vote("A"))
print(vote("B"))
print(vote("A"))
```

```python
def fact(n):
  if n==0:
    return 1
  else:
    return n*fact(n-1)
print(fact(5))

#recursive function
```

```python
#fibonacci series
 #[0,1,1,2,3,5]

def fibonacci(n):
  if n<=1:
    return n
  else:
    return fibonacci(n-1)+fibonacci(n-2)
for i in range(6):
  print(fibonacci(i))
```

```python
data = {
    "Name": [
        "Asha Hassan", "Dilli Babu N", "Ekanth Reddy T",
        "Parameshwar D", "Preethi B V", "Shubrati",
        "Sravanthi Veluru", "Supriya Moudgal"
    ],
    "Age": [24, 29, 35, 40, 27, 31, 36, 28],
    "Salary": [48000, 55000, 72000, 80000, 60000, 51000, 75000, 64000],
    "Department": ["HR", "IT", "Finance", "IT", "HR", "Finance", "IT", "HR"]
}

df = pd.DataFrame(data)
print(df)
```
