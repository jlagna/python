```python
import math
class circle:
  def __init__(self,r):
    self.r=r
  def area(self):
    r=10
    return r*self.r*math.pi
  def cir(self):
    return 2*math.pi*self.r
```

```python
class test:
  def __init__(self,n):
    self.n=n

  def compute(self,x=None):
    if x is None:
      x = self.n
    return 1 if x==0 else x * self.compute(x-1)

f=test(5)
print(f.compute())
```

```python
class filepath:
  def __init__(self,path):
    self.path=path
  def return_files(self,input_value):
    if input_value=='csv':
      return glob.glob(self.path+"/*.csv")
    elif input_value=='pdf':
      return glob.glob(self.path+"/*.pdf")
    elif input_value=='html':
     return glob.glob(self.path+"/*.html")
    elif input_value=='png':
     return glob.glob(self.path+"/*.png")
```

```python
object1=filepath("/content")
object1.return_files("html")
```

```python
class Person:
  def speak(self):
    return "i am studying"

class student(Person):
  def study(self):
    return "i am studying"

s=student()
print(s.speak())
print(s.study())
s.study()
```

```python
class Person:
  def speak(self):
    return "i can speak"

class student(Person):
  def student(self):
    return "i am studying"

s=student()
s.student()
```

```python
class vehicle:
  def move(self):
    return "moving on road"

class car(vehicle):
  def fuel(self):
    return "runs on petrol"

class electriccar(car):
  def battery(self):
    return "powered by electricity"

ec=electriccar()
print(ec.move())
print(ec.fuel())
print(ec.battery())

```
