```python
def order_details(**kwargs):
  print("order details:",kwargs)

order_details(item="laptop",price=50000,quantity=2)
```

```python
def order_details(**kwargs):
  print("order details:",kwargs['price'])

order_details(item="laptop",price=50000,quantity=2)
```

```python
def profile(id,*args,**kwargs):
  print(f"ID:{id}")
  print("Args:",args)
  print("kwargs:",kwargs)

profile(101,"python","azure",location="bangalore",role="trainer")
```

```python
def greet(**kwargs):
  for key, value in kwargs.items():   #keyvalue
    print(f"{key}={value}")

greet(name="bharath",role="data engineer",company="edukron")
```

```python
def introduce(name,**kwargs):
  print("name:",name)
  if "age" in kwargs:
    print("age:",kwargs["age"])
  if "job" in kwargs:
    print("job:",kwargs["job"])

introduce("anusha",age=25,job="data analyst")
```

```python
def introduce(name,**kwargs):
  print("name:",name)
  if "age" in kwargs:
    print("age:",kwargs["age"])
  if "job" in kwargs:
    print("job:",kwargs["job"])
    #key should unique in dictionary

introduce("anusha",age=25,age=35,job="data analyst")
```

```python
def greet(*args,**kwargs):
    for person in args:
      print("Hello",person)
    for k, v in kwargs.items():
      print(f"{k}:{v}")

greet("Anusha","kiran",course="python",batch="morning")
```

```python
def calculate(*args,**kwargs):
  total = sum(args)
  print("total:",total)
  if "discount" in kwargs:
    total -=kwargs["discount"]
  print("Final:",total)

calculate(100,200,50,discount=30)
```

```python
words=["apple","banana","apricot","cherry"]

starts_with =lambda w,**kwargs:w.startswith(kwargs.get("prefix","a"))

result=list(filter(lambda w: starts_with(w,prefix="ap"),words))
print(result)
```
